```python
from langchain.agents import create_agent
from dotenv import load_dotenv
import os
from langchain.tools import tool

load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY") # pyright: ignore[reportArgumentType]
```

```python

@tool 
def querySqlDB(query:str)->str:
    """This function is used to query SQL database. Input - valid SQL Query which be queried directly"""
    print(query)
    return "132"


model = create_agent(model="google_genai:gemini-3.6-flash", system_prompt="You are database query manager, we have SQL db and users table. Create valid queries.", tools=[querySqlDB])
```

```python
response = model.invoke(({
    "messages": [
        {"role": "user", "content": "Tell me how many users we have currently"}
    ]
}))
print(response)
```
