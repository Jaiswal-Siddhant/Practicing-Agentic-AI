## Roles: 
There are three core roles: `system`, `user`, `assistant`

| Role        | Comes from            | Purpose                                        |
| ----------- | --------------------- | ---------------------------------------------- |
| `system`    | Developer/application | Defines how the AI should behave               |
| `user`      | User                  | Gives the actual request/task                  |
| `assistant` | AI/model              | Represents the AI's previous responses/actions |

A simple conversation looks like:
```md
    SYSTEM:
    You are a helpful coding assistant.
    Always explain code clearly.

    USER:
    How do I reverse a string in Python?

    ASSISTANT:
    You can use string slicing:

    text[::-1]
```

```
SystemMessage
    ↓
How should the model behave?

HumanMessage
    ↓
What does the human want?

AIMessage
    ↓
What did the model say/do?

ToolMessage
    ↓
What did the external tool return?
```

````python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import (
    SystemMessage,
    HumanMessage
)

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

messages = [
    SystemMessage(
        content="You are a helpful Python tutor."
    ),

    HumanMessage(
        content="Explain list comprehensions in 2 lines"
    )
]

response = model.invoke(messages)

print(response.text)
````

````python
messages.append(response)

messages.append(
    HumanMessage(
        content="Give me a practical example."
    )
)

response = model.invoke(messages)

print(response.content)
````
